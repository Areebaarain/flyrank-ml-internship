## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*
### Data limits

This data can never tell me the true causal reason why a piece of content is underperforming. The history may be unbalanced, so some content or clients may have more observations than others. Early rows may rely on GSC-only data, which can make comparisons less consistent across the full history. Time windows can also overlap, so I cannot always treat neighboring observations as fully independent. The data can identify patterns and refresh opportunities, but it cannot prove that refreshing a piece of content will definitely improve its future performance.
